```python
import pandas as pd
import scipy.io

# 1. Load the .mat file
mat_data = scipy.io.loadmat('data/EMG-data (1).mat')

# 2. Filter out the MATLAB metadata keys
# This leaves only your actual variables/matrices
data_keys = [key for key in mat_data.keys() if not key.startswith('__')]

# 3. Convert to a Pandas DataFrame
# Case A: If you have a single 2D matrix you want as a dataframe
matrix_name = data_keys[0] # Takes the first available data variable
df = pd.DataFrame(mat_data[matrix_name])

# Case B: If your .mat file contains multiple 1D arrays (columns)
# df = pd.DataFrame({key: mat_data[key].flatten() for key in data_keys})

print(df.head())
```
